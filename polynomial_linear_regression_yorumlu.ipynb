```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("/Users/berkecelik/Downloads/poly.csv")
data.head()
```

```python
# Bağımsız değişkeni (x) veri setinin ilk sütunundan alıyoruz
# reshape(-1,1) ile x'i tek sütunlu 2 boyutlu bir diziye dönüştürüyoruz (sklearn bu formatı bekler)
# Bağımlı değişkeni (y) veri setinin ikinci sütunundan aynı şekilde alıyoruz
x = data.iloc[:,0].values.reshape(-1,1)
y = data.iloc[:,1].values.reshape(-1,1)
```

```python
# Doğrusal regresyon (linear regression) modelini sklearn'den içe aktarıyoruz
# LinearRegression nesnesi oluşturuyoruz
# fit() ile modeli x ve y verileri üzerinde eğitiyoruz (basit doğrusal model)
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x,y)
```

```python
# Gerçek veri noktalarını (x,y) saçılım grafiği (scatter plot) olarak çiziyoruz
# Doğrusal modelin tahminlerini kırmızı bir çizgi olarak grafiğe ekliyoruz
# Bu grafik doğrusal modelin veriye ne kadar iyi uyduğunu görsel olarak gösterir
plt.scatter(x,y)
plt.plot(x,lr.predict(x), color="red")
```

```python
# Polinomsal özellik dönüşümü için PolynomialFeatures sınıfını içe aktarıyoruz
# degree=3 ile x değişkeninden 3. dereceye kadar polinom terimleri (x, x^2, x^3) üretecek nesneyi oluşturuyoruz
from sklearn.preprocessing import PolynomialFeatures
pr = PolynomialFeatures(degree = 3)
```

```python
# x verisini polinomsal özelliklere dönüştürüyoruz (x, x^2, x^3 sütunlarını içeren yeni bir veri seti oluşuyor)
xpr = pr.fit_transform(x)
```

```python
# Polinomsal özellikler üzerinde eğitilecek ikinci bir doğrusal regresyon modeli oluşturuyoruz
# (Not: Polinom terimlerine doğrusal katsayılar uygulandığı için hala 'linear regression' kullanılıyor)
lr2 = LinearRegression()
```

```python
# İkinci modeli (lr2) polinomsal olarak dönüştürülmüş xpr verisi ve y hedef değeri ile eğitiyoruz
lr2.fit(xpr,y)
```

```python
# Eğitilen polinomsal regresyon modeliyle xpr verisi için tahminler üretiyoruz
yhead = lr2.predict(xpr)
```

```python
# Gerçek veri noktalarını tekrar saçılım grafiği olarak çiziyoruz
# Basit doğrusal modelin tahminlerini kırmızı çizgiyle gösteriyoruz
# Polinomsal modelin tahminlerini yeşil çizgiyle gösteriyoruz
# Böylece iki modelin veriye uyumu karşılaştırılabiliyor (polinomsal model genelde daha iyi uyum sağlar)
plt.scatter(x,y)
plt.plot(x, lr.predict(x), color="red")
plt.plot(x, lr2.predict(xpr), color="green")
```

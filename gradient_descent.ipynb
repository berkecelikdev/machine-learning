```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([1.0, 2.0, 3.0])
y_train = np.array([2.0, 4.0, 6.0])
```

```python
def compute_cost(x, y, w, b):
    m = x.shape[0]
    cost = 0

    for i in range(m):
        f_wb = w * x[i] + b
        cost = cost + (f_wb - y[i]) ** 2
    total_cost = 1 / (2 * m) * cost

    return total_cost
```

```python
def compute_gradient(x, y, w, b):
    m = x.shape[0]
    dj_dw = 0
    dj_db = 0

    for i in range(m):
        f_wb = w * x[i] + b
        dj_dw_i = (f_wb - y[i]) * x[i]
        dj_db_i = f_wb - y[i]
        dj_dw += dj_dw_i
        dj_db += dj_db_i

    dj_dw = dj_dw / m
    dj_db = dj_db / m

    return dj_dw, dj_db
```

```python
w = 1.0
b = 1.0

cost = compute_cost(x_train, y_train, w, b)
dj_dw, dj_db = compute_gradient(x_train, y_train, w, b)

print(f"Cost: {cost}")
print(f"dj_dw: {dj_dw}, dj_db: {dj_db}")
```

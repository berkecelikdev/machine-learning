```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("/Users/berkecelik/Downloads/ürün.csv")
data.head()
```

```python
# Bağımsız değişkenleri (x) veri setinin ilk iki sütunundan (yaş, maaş) alıyoruz
# Bağımlı/hedef değişkeni (y) üçüncü sütundan (satınalma: 0 ya da 1) alıp tek sütunlu diziye dönüştürüyoruz
x = data.iloc[:,0:2].values
y = data.iloc[:,2].values.reshape(-1,1)
```

```python
# Satın almayanları (satinalma == 0) ve satın alanları (satinalma == 1) ayrı iki gruba ayırıyoruz
# Bu ayrım, sınıfları görselleştirmede kullanılacak
zeros = data[data.satinalma == 0]
ones = data[data.satinalma == 1]
```

```python
# Satın almayan müşterileri (yaş, maaş) kırmızı noktalarla çiziyoruz
# Satın alan müşterileri (yaş, maaş) yeşil noktalarla çiziyoruz
# Bu grafik iki sınıfın veri uzayında nasıl dağıldığını görsel olarak inceler
plt.scatter(zeros.yaş, zeros.maaş, color = "red")
plt.scatter(ones.yaş, ones.maaş, color = "green")
```

```python
# Veri setini eğitim (train) ve test setlerine ayırmak için train_test_split fonksiyonunu içe aktarıyoruz
# Verinin %30'unu test, %70'ini eğitim seti olarak ayırıyoruz (random_state ile sonuçlar tekrarlanabilir olur)
from sklearn.model_selection import train_test_split
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size = 0.3, random_state = 44)
```

```python
# Özellikleri (x) standartlaştırmak için StandardScaler sınıfını içe aktarıyoruz
# StandardScaler nesnesi oluşturuyoruz (yaş ve maaş farklı ölçeklerde olduğu için ölçeklendirme gerekir)
from sklearn.preprocessing import StandardScaler
std_scaler = StandardScaler()
```

```python
# Eğitim verisini standartlaştırıyoruz (ortalama=0, standart sapma=1 olacak şekilde ölçekliyoruz) ve öğrenilen parametreleri kaydediyoruz
# Test verisini ise eğitim verisinden öğrenilen aynı ölçekle dönüştürüyoruz (veri sızıntısını önlemek için sadece transform kullanılır)
xtrain1 = std_scaler.fit_transform(xtrain)
xtest1 = std_scaler.transform(xtest)
```

```python
# Lojistik regresyon (logistic regression) sınıflandırma modelini sklearn'den içe aktarıyoruz
# LogisticRegression nesnesi oluşturuyoruz
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
```

```python
# Modeli, standartlaştırılmış eğitim verileri (xtrain1) ve hedef değerler (ytrain) ile eğitiyoruz
model.fit(xtrain1, ytrain)
```

```python
# Eğitilen model ile standartlaştırılmış test verileri (xtest1) üzerinde tahmin yapıyoruz
y_predict = model.predict(xtest1)
```

```python
# Test verisi için üretilen tahminleri (0 ya da 1) ekrana yazdırıyoruz
y_predict
```

```python
# Modelin test seti üzerindeki doğruluk (accuracy) skorunu hesaplıyoruz
model.score(xtest1, ytest)
```

```python
# Karışıklık matrisi (confusion matrix) hesaplamak için confusion_matrix fonksiyonunu içe aktarıyoruz
# Gerçek değerler (ytest) ile tahmin edilen değerleri (y_predict) karşılaştırarak karışıklık matrisini oluşturuyoruz
# Matris; doğru/yanlış pozitif ve negatif tahmin sayılarını gösterir
from sklearn.metrics import confusion_matrix
conf_matrix = confusion_matrix(ytest, y_predict)
conf_matrix
```
